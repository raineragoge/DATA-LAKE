```python
# Creamos atajos para hacer que sea mucho más fácil y flexible trabajar.
# Además que es más fácil realizar futuros cambios.
PATH = "/content/data_lake/procesed/"
PATH_RAW = "/content/data_lake/raw/"
```

```python
# Creamos todas las carpetas que vamos autilizar en el ejercicio.
import os

carpetas = [
    "/content/data_lake",
    "/content/data_lake/procesed",
    "/content/data_lake/raw",
    "/content/data_lake/COMPLETADO"
]

for carpeta in carpetas:
    os.makedirs(carpeta, exist_ok=True)

print("Carpetas creadas correctamente.")
```

```python
# Leemos los archivos descargados previamente.
import pandas as pd

clientes = pd.read_csv(PATH_RAW + "customers.csv")

print(clientes)
```

```python
# Leemos los archivos descargados previamente.
import pandas as pd

productos = pd.read_json(PATH_RAW + "productos.json")

print(productos)
```

```python
# Limpiamos los archivos para que sean leibles por el programa. (cambiomos los "|" por ",")
import pandas as pd

pedidos = pd.read_csv(
    PATH_RAW + "orders.txt",
    sep="|",
    header=None,
    names=["order_id", "customer_id", "product_id", "quantity"]
)

pedidos.to_csv("orders.csv", index=False)
```

```python
# Leemos los archivos descargados previamente.
import pandas as pd

productos = pd.read_json(PATH_RAW + "productos.json")

productos.to_csv("productos.csv", index=False)

print(productos)
```

```python
# Unir pedidos con clientes por customer_id
datos = pedidos.merge(clientes, on="customer_id", how="left")

# Unir el resultado con productos por product_id
datos = datos.merge(productos, on="product_id", how="left")

# Guardar la tabla completa
datos.to_csv("filtadro.csv", index=False)

print(datos)
```

```python
# Guardar la tabla completa
datos.to_csv(PATH +"tabla_completa.csv", index=False)

print(datos)
```

```python
#Ejecutar la operación para saber los precios totales reales.
datos["total"] = datos["quantity"] * datos["price"]
```

```python
#Guardar el CSV con la nueva columna
datos.to_csv(PATH +"total_sin_iva.csv", index=False)
print(datos)
```

```python
#Ejecutar la operación para saber los precios totales reales con iva
datos["con_iva"] = datos["total"] + ((datos["total"] * 0.21))
```

```python
#Guardar el CSV con la nueva columna
datos.to_csv(PATH + "total_con_iva.csv", index=False)
print(datos)
```

```python
# Eliminar columna city
df = pd.DataFrame(datos)
datos = df.drop(['city'], axis=1)
```

```python
#Guardar el CSV con la nueva columna
datos.to_csv(PATH + "eliminar_city.csv", index=False)
print(datos)
```

```python
# Edad mayor que 30:
df.loc[df["age"] >= 30]
```

```python
# Edad menor que 30:
df.loc[df["age"] < 30]
```

```python
# Ordenar los datos según el valor de IVA
datos_ordenados = datos.sort_values(by="con_iva")
```

```python
# Ordenar los datos según el valor de IVA de mayor a menor
datos_ordenados = datos.sort_values(by="con_iva", ascending=False)
```

```python
datos_ordenados.to_csv(PATH + "datos_ordenados.csv", index=False)
print(datos_ordenados)
```

```python
#Guardar los datos finales completados:
datos_ordenados.to_csv("/content/data_lake/COMPLETADO/completado.csv", index=False)
print(datos_ordenados)
```
