```python
import pandas as pd

pedidos = pd.read_csv(
    PATH_RAW + "orders.txt",
    sep="|",
    header=None,
    names=["order_id", "customer_id", "product_id", "quantity"]
)

pedidos.to_csv("orders.csv", index=False)
```

```python
import pandas as pd

productos = pd.read_json(PATH_RAW + "productos.json")

productos.to_csv("productos.csv", index=False)

print(productos)
```

```python
# Unir pedidos con clientes por customer_id
datos = pedidos.merge(clientes, on="customer_id", how="left")

# Unir el resultado con productos por product_id
datos = datos.merge(productos, on="product_id", how="left")

# Guardar la tabla completa
datos.to_csv("filtadro.csv", index=False)

print(datos)
```

```python
# Guardar la tabla completa
datos.to_csv(PATH +"tabla_completa.csv", index=False)

print(datos)
```

```python
datos["total"] = datos["quantity"] * datos["price"]
```

```python
 #Guardar el CSV con la nueva columna
datos.to_csv(PATH +"total_sin_iva.csv", index=False)
print(datos)
```

```python
datos["con_iva"] = datos["total"] + ((datos["total"] * 0.21))
```

```python
#Guardar el CSV con la nueva columna
datos.to_csv(PATH + "total_con_iva.csv", index=False)
print(datos)
```

```python
df = pd.DataFrame(datos)
datos = df.drop(['city'], axis=1)
```

```python
#Guardar el CSV con la nueva columna
datos.to_csv(PATH + "eliminar_city.csv", index=False)
print(datos)
```

```python
# Edad mayor que 30:
df.loc[df["age"] >= 30]
```

```python
datos_ordenados = datos.sort_values(by="con_iva")
```

```python
datos_ordenados = datos.sort_values(by="con_iva", ascending=False)
```

```python
datos_ordenados.to_csv(PATH + "datos_ordenados.csv", index=False)
print(datos_ordenados)
```

```python
#Guardar los datos finales completados:
datos_ordenados.to_csv("/content/data_lake/COMPLETADO/completado.csv", index=False)
print(datos_ordenados)
```
